```python

import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-5, 11)

#unit impulse
delta = (n == 0).astype(float)

#unit step
step = (n >= 0).astype(float)

#unit ramp
ramp = n * step

#exponential
expo = 0.8**n * step

fig, axes = plt.subplots(1, 4, figsize=(13, 3))

signals = [delta, step, ramp, expo]

titles = [
    "Unit Impulse",
    "Unit Step",
    "Unit Ramp",
    "Exponential"
]

for ax, x, title in zip(axes, signals, titles):
  ax.stem(n, x)
  ax.set_title(title)
  ax.set_xlabel("n")
  ax.set_ylabel("Amplitude")
  ax.grid()

  plt.tight_layout()
  plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 11)

values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(12, 3))

for ax, a in zip(axes, values):

    x = (a**n)

    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 41)

x1 = np.cos(2*np.pi*n/8)
x2 = np.cos(2*np.pi*n/7.5)

plt.figure(figsize=(10, 4))

plt.stem(n, x1, linefmt='b-', markerfmt='bo',
         basefmt=' ', label='Period 8')

plt.plot(n, x2, 'r--o', label='Period 7.5')

plt.xlabel("n")
plt.ylabel("Amplitude")
plt.title("Comparison of Two Signals")
plt.legend()
plt.grid()
plt.show()
```

```python
from scipy.io import wavfile
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display

fs, x = wavfile.read("my_voice.wav")

if x.ndim > 1:
    x = x.mean(axis=1)

x = x / np.max(np.abs(x))

t = np.arange(len(x)) / fs

print("Sampling rate:", fs, "Hz")
print("Duration:", len(x)/fs, "seconds")
print("Number of samples:", len(x))
```

```python
plt.plot(t, x)
plt.xlabel("time (s)")
plt.ylabel("amplitude")
plt.title("My recording")
plt.show()
```
